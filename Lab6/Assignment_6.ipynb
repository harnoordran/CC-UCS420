## Q1. Sensor Readings

A sensor has an average measurement error of **+2°C**.

**a)** Add 2°C to every reading using vectorization.<br>
**b)** Convert all temperatures from Celsius to Fahrenheit using **F = (9/5)C + 32**.<br>
**c)** Identify all readings greater than 32°C using Boolean indexing.<br>
**d)** Count how many readings exceed 32°C.<br>
**e)** Explain how vectorization and Boolean indexing are more efficient than explicitly iterating through every value.

````python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a) Add 2°C using vectorization
corrected_temperature = temperature + 2
print("a) Corrected temperatures:", corrected_temperature)

# b) Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("b) Fahrenheit:", fahrenheit)

# c) Readings greater than 32°C
readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("c) Readings greater than 32°C:", readings_greater_32)

# d) Count readings greater than 32°C
count_greater_32 = np.sum(corrected_temperature > 32)
print("d) Count:", count_greater_32)
````

### e) Explanation

**Vectorization** performs an operation on the complete NumPy array at once, so we do not need a `for` loop for every reading.

**Boolean indexing** creates a True/False condition for each element and directly selects the elements satisfying the condition. This makes array processing concise and generally faster than explicit Python loops.

## Q2. Daily Steps

**a)** Calculate total steps.<br>
**b)** Calculate mean steps.<br>
**c)** Find maximum and minimum.<br>
**d)** Calculate total steps for each day using `axis=0`.<br>
**e)** Calculate total steps for each user using `axis=1`.<br>
**f)** Find the user/day position containing the maximum using `argmax()`.

````python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a) Total steps:", np.sum(steps))
print("b) Mean steps:", np.mean(steps))
print("c) Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))
print("d) Total for each day:", np.sum(steps, axis=0))
print("e) Total for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("f) Maximum position (user index, day index):", max_position)
print("   Maximum value:", steps[max_position])
````

## Q3. Slicing, Copying, Reshaping and Flattening



````python
# a)
original = np.array([1, 2, 3, 4, 5, 6])
print("a) Original:", original)

# b)
subset = original[1:5]
print("\nb) Subset:", subset)

# c) A slice is a view
subset[0] = 999
print("\nc) After modifying subset:")
print("Original:", original)
print("Subset:", subset)

# d) copy() creates an independent array
copied_array = original[1:5].copy()
copied_array[0] = 500
print("\nd) After modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

# e)
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne) 3 x 4 matrix:")
print(matrix)

# f)
print("\nf) First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g)
flattened = matrix.flatten()
raveled = matrix.ravel()
print("\ng) flatten():", flattened)
print("ravel():", raveled)

# h)
raveled[0] = 1000
print("\nh) After changing ravel():")
print("Raveled:", raveled)
print("Original matrix:")
print(matrix)

# i)
flattened[1] = 2000
print("\ni) After changing flatten():")
print("Flattened:", flattened)
print("Original matrix:")
print(matrix)

# j)
print("\nj) Shape:", matrix.shape)
print("ndim:", matrix.ndim)
print("size:", matrix.size)
print("dtype:", matrix.dtype)
````

### Observation

- A normal slice is a **view**, so changing it can change the original array.
- copy() creates an independent copy.
- ravel() generally returns a view when possible, so changing it can affect the original.
- flatten() returns a copy, so changing it does not affect the original.

## Q4. Cognitive-Assistive System and OLS

Given:
```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

Each row represents one user and each column represents a feature: **sleep hours, activity level, stress level**.



````python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a)
print("a) Shape:", X.shape)
print("Dimensions:", X.ndim)

# b)
X_T = X.T
print("\nb) X.T:")
print(X_T)

# c)
XTX = X_T @ X
print("\nc) X.T @ X:")
print(XTX)

# d)
XTX_inv = np.linalg.inv(XTX)
print("\nd) Inverse:")
print(XTX_inv)

# e)
beta = XTX_inv @ X_T @ y
print("\ne) OLS coefficients:")
print(beta)

# g)
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\ng) New user:", new_user)
print("Predicted assistance score:", predicted_score)
````

### f) Interpretation of coefficients

The three coefficients correspond respectively to:

1. **Sleep hours**
2. **Activity level**
3. **Stress level**

They represent the change in the predicted assistance score associated with a one-unit change in the corresponding feature, while the other features are held constant.
