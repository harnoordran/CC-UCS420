# Q6. Employee Dataset

```python
employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)

employees
```

```python
employees.to_csv("employees.csv", index=False)
print("Employees.csv created successfully.")
```

## Q6(e). Calculate Statistics

The following statistics are calculated:
1. Average salary
2. Total bonus paid
3. Youngest employee's age
4. Highest performance rating

```python
avg_salary = employees["Salary"].mean()
print("Average Salary: ", avg_salary)

total_bonus = employees["Bonus"].sum()
print("Total Bonus: ", total_bonus)

youngest_age = employees["Age"].min()
print("Youngest Age: ", youngest_age)

highest_rating= employees["Rating"].max()
print("Highest Rating: ", highest_rating)
```

## Q6(f). Sort by Salary

The DataFrame is sorted according to Salary in descending order.

```python
sort_employees = employees.sort_values(by="Salary", ascending=False)
sort_employees
```

## Q6(g). Categorize Employees Based on Rating

```python
def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"

    elif rating >= 4.0:
        return "Good"

    else:
        return "Average"

employees["Performace_Category"] = employees["Rating"].apply(categorize_rating)

employees

    
```

## Q6(j). Filter Employees

```python
print("Employees with more than 5 years of experience: ")
display(employees[employees["Years_of_Experience"] > 5] )

print("Employees in IT Department: ")
display(employees[employees["Department"] == "IT"])
```
