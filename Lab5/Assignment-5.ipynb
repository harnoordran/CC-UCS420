## Q2. Questions on Basic NumPy Array

### a) Reverse the NumPy array


```python
arr = np.array([1, 2, 3, 6, 4, 5])

reversed_arr = arr[::-1]

print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

### b) Find the most frequent value and its index/indices



```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
val_x, counts_x = np.unique(x, return_counts=True)
most_freq_x = val_x[np.argmax(counts_x)]
indices_x = np.where(x == most_freq_x)[0]
print(most_freq_x)
print(indices_x)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
val_y, counts_y = np.unique(y, return_counts=True)
most_freq_y_vals = val_y[counts_y == counts_y.max()]
for val in most_freq_y_vals:
    indices_y = np.where(y == val)[0]
    print(val)
    print(indices_y)
```

## Q4. Create a 1-D NumPy array with 25 evenly spaced numbers

Create an array named **Harnoor** with evenly spaced 25 numbers from 10 to 100 using `linspace()`.

Print:
- Dimensions of the array
- Shape
- Total number of elements
- Data type of each element
- Total number of bytes consumed

Then find the transpose using `reshape()`.

**Can the same be done using the `T` attribute?**

```python
Harnoor = np.linspace(10, 100, 25)

print("Array:")
print(Harnoor)

print("\nDimensions:", Harnoor.ndim)
print("Shape:", Harnoor.shape)
print("Total elements:", Harnoor.size)
print("Data type:", Harnoor.dtype)
print("Total bytes:", Harnoor.nbytes)

# Transpose using reshape
transpose_reshape = Harnoor.reshape(25, 1)
print(transpose_reshape)

# Transpose using T
print(Harnoor.T)
```

## Q5. Create and perform operations on a 2-D NumPy array

Create a **2-D array of 3 rows and 4 columns** named `ucs420_Harnoor` with:

Compute:
- Mean
- Median
- Maximum
- Minimum
- Unique elements

Then:
1. Reshape the array to **4 rows and 3 columns** and name it `reshaped_ucs420_Harnoor`.
2. Resize the array to **2 rows and 3 columns** and name it `resized_ucs420_Harnoor`.

```python
ucs420_harnoor = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print(np.mean(ucs420_harnoor))
print(np.median(ucs420_harnoor))
print(np.max(ucs420_harnoor))
print(np.min(ucs420_harnoor))
print(np.unique(ucs420_harnoor))

reshaped_ucs420_harnoor = ucs420_harnoor.reshape(4, 3)
print(reshaped_ucs420_harnoor)

reshaped_ucs420_harnoor = np.resize(ucs420_harnoor,(2, 3))
print(reshaped_ucs420_harnoor)
```
